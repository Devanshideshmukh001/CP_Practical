```python
#if elif else statement for marks of students
#above 90 marks = A grade, 80-89 marks = B grade, 70-79 marks = C grade, 60-69 marks = D grade, 50-59 marks = E grade, below 50 marks = F grade
marks = int(input('Enter your marks: '))
if marks >= 90:
    print('You scored grade A')
elif marks >= 80:
    print('You scored grade B')
elif marks >= 70:
    print('You scored grade C')
elif marks >= 60:
    print('You scored grade D')
elif marks >= 50:
    print('You scored grade E')
else:
    print('You scored grade F')
```
